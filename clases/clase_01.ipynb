# Guía completa de Git y GitHub

Una explicación paso a paso, pensada para personas que nunca han usado control de versiones.

---

## 1. El problema que resuelven

Imagina que estás escribiendo un trabajo importante y vas guardando copias así:

```
informe.docx
informe_v2.docx
informe_v2_corregido.docx
informe_final.docx
informe_final_AHORA_SI.docx
```

Esto genera varios problemas:

- No sabes con certeza qué cambió entre una versión y otra.
- Si trabajas con otra persona, es muy fácil que uno sobrescriba el trabajo del otro.
- Si algo se rompe, cuesta mucho volver a una versión que funcionaba.

**Git** y **GitHub** existen para resolver exactamente esto, especialmente cuando se trabaja con código.

---

## 2. Qué es Git y qué es GitHub

Son dos cosas distintas que suelen confundirse.

| | Git | GitHub |
|---|---|---|
| Qué es | Un programa | Un sitio web (un servicio en internet) |
| Dónde funciona | En tu computador | En la nube |
| Para qué sirve | Registrar el historial de cambios de un proyecto | Guardar proyectos de Git en internet, compartirlos y colaborar |
| Necesita internet | No | Sí |

Una analogía útil:

- **Git** es como la cámara de fotos con la que vas sacando fotografías de tu proyecto a lo largo del tiempo.
- **GitHub** es como un álbum en la nube donde subes esas fotografías para guardarlas y compartirlas con otras personas.

Puedes usar Git sin GitHub. Pero GitHub sin Git no tiene mucho sentido, porque lo que guarda son proyectos de Git.

Existen alternativas a GitHub que funcionan igual, como GitLab y Bitbucket.

## 3. Conceptos fundamentales

Antes de escribir comandos, conviene entender cinco ideas.

### 3.1 Repositorio

Un **repositorio** (o "repo") es una carpeta de proyecto que Git está vigilando. Además de tus archivos, contiene una carpeta oculta llamada `.git` donde se guarda todo el historial.

- **Repositorio local**: el que está en tu computador.
- **Repositorio remoto**: una copia del mismo proyecto guardada en internet, por ejemplo en GitHub.

### 3.2 Commit

Un **commit** es una fotografía del proyecto en un momento determinado. Cada commit guarda:

- Qué archivos cambiaron y cómo.
- Quién hizo el cambio.
- Cuándo se hizo.
- Un mensaje que explica por qué se hizo.

El historial de un proyecto no es más que una lista de commits, uno detrás de otro.

### 3.3 Las tres zonas de Git

Este es el concepto que más cuesta al principio, y el más importante. Git organiza tu trabajo en tres zonas:

```
  Directorio de trabajo   -->   Área de preparación   -->   Repositorio
  (donde editas)                (staging area)              (historial guardado)
                     git add                     git commit
```

1. **Directorio de trabajo**: tus archivos tal como los ves y editas.
2. **Área de preparación (staging)**: una especie de "sala de espera" donde eliges qué cambios van a entrar en la próxima fotografía.
3. **Repositorio**: donde quedan guardados los commits de forma permanente.

Una analogía: estás armando una caja para enviar por correo.

- El directorio de trabajo es tu casa, con todas tus cosas.
- El área de preparación es la caja abierta, donde vas poniendo lo que quieres enviar.
- El commit es cerrar la caja, pegarle una etiqueta y despacharla.

¿Por qué existe la zona intermedia? Porque a veces cambias diez archivos pero solo quieres guardar tres en un commit, y los otros siete en otro commit distinto. Así el historial queda ordenado.

### 3.4 Rama (branch)

Una **rama** es una línea de trabajo independiente. Permite probar cosas nuevas sin afectar la versión principal.

```
main:        A --- B --- C ------------- F
                          \             /
nueva-func:                D --- E ----
```

- La rama principal suele llamarse `main` (en proyectos antiguos, `master`).
- Desde `C` se creó la rama `nueva-func`, donde se hicieron los commits `D` y `E`.
- Cuando el trabajo estuvo listo, se unió de vuelta a `main` en `F`.

Analogía: es como hacer una copia de un documento para probar una redacción distinta. Si te gusta, la incorporas al original. Si no, la descartas y el original queda intacto.

### 3.5 Merge (fusión)

Un **merge** es la acción de unir los cambios de una rama con otra. En el diagrama anterior, `F` es el commit de merge.

## 4. Instalación y configuración inicial

### 4.1 Instalar Git

- **Windows**: descargar desde https://git-scm.com e instalar con las opciones por defecto. Se instala también "Git Bash", una terminal para usar Git.
- **macOS**: abrir la Terminal y escribir `git --version`. Si no está instalado, el sistema ofrecerá instalarlo.
- **Linux (Ubuntu/Debian)**: `sudo apt install git`

Para comprobar que funciona:

```bash
git --version
```

### 4.2 Decirle a Git quién eres

Esto se hace una sola vez por computador. Git usa estos datos para firmar tus commits.

```bash
git config --global user.name "Tu Nombre"
git config --global user.email "tu-correo@ejemplo.com"
```

Usa el mismo correo con el que crearás tu cuenta de GitHub.

Para que la rama principal se llame `main` por defecto:

```bash
git config --global init.defaultBranch main
```

## 5. Primeros pasos con Git (trabajo local)

### 5.1 Crear un repositorio

Entra a la carpeta de tu proyecto desde la terminal y escribe:

```bash
git init
```

Desde ese momento, Git empieza a vigilar esa carpeta.

### 5.2 Ver el estado del proyecto

El comando que más vas a usar:

```bash
git status
```

Te dice qué archivos cambiaron, cuáles están en el área de preparación y cuáles Git todavía no conoce. Ante cualquier duda, ejecuta `git status`.

### 5.3 Preparar cambios

```bash
git add archivo.py          # agrega un archivo específico
git add carpeta/            # agrega una carpeta completa
git add .                   # agrega todos los cambios de la carpeta actual
```

### 5.4 Guardar un commit

```bash
git commit -m "Agrega función para calcular el promedio"
```

El texto entre comillas es el mensaje del commit. Debe explicar de forma breve qué hace el cambio.

Buenos mensajes:

- `Corrige error al dividir por cero en promedio`
- `Agrega formulario de inscripción`

Malos mensajes:

- `cambios`
- `asdf`
- `arreglado`

### 5.5 Ver el historial

```bash
git log                 # historial completo
git log --oneline       # historial resumido, una línea por commit
```

Cada commit tiene un identificador único, una cadena como `a3f9c21`. Sirve para referirse a ese commit específico.

### 5.6 Ver qué cambió

```bash
git diff                # cambios que aún no están en el área de preparación
git diff --staged       # cambios que ya están preparados para el commit
```

### 5.7 Un ejemplo completo

```bash
mkdir mi-proyecto
cd mi-proyecto
git init

# se crea un archivo y se escribe algo en él
echo "print('Hola mundo')" > hola.py

git status              # Git muestra hola.py como archivo nuevo, sin seguimiento
git add hola.py
git status              # ahora aparece como listo para el commit
git commit -m "Primer programa: saludo"
git log --oneline       # aparece el primer commit
```

## 6. Trabajar con ramas

### 6.1 Comandos básicos

```bash
git branch                      # lista las ramas; la actual aparece marcada con *
git branch nueva-funcion        # crea una rama nueva
git switch nueva-funcion        # se cambia a esa rama
git switch -c otra-rama         # crea la rama y se cambia a ella en un solo paso
```

En tutoriales antiguos verás `git checkout` en lugar de `git switch`. Hacen lo mismo en este contexto; `switch` es más nuevo y más claro.

### 6.2 Unir una rama con main

```bash
git switch main                 # primero vuelves a la rama que recibirá los cambios
git merge nueva-funcion         # traes los cambios de la otra rama
git branch -d nueva-funcion     # opcional: borras la rama que ya no necesitas
```

### 6.3 Conflictos

Un **conflicto** ocurre cuando dos ramas modificaron la misma parte del mismo archivo y Git no sabe cuál versión conservar. No es un error grave: Git simplemente te pide que decidas tú.

El archivo afectado quedará marcado así:

```
<<<<<<< HEAD
print("Hola")
=======
print("Buenos días")
>>>>>>> nueva-funcion
```

- Lo que está entre `<<<<<<< HEAD` y `=======` es la versión de la rama donde estás.
- Lo que está entre `=======` y `>>>>>>>` es la versión que viene de la otra rama.

Para resolverlo:

1. Abre el archivo y deja el contenido como debe quedar finalmente (puede ser una versión, la otra o una mezcla).
2. Borra las líneas con `<<<<<<<`, `=======` y `>>>>>>>`.
3. Guarda el archivo y ejecuta:

```bash
git add archivo.py
git commit -m "Resuelve conflicto en saludo"
```

## 7. GitHub: llevar el proyecto a internet

### 7.1 Crear una cuenta y un repositorio

1. Crear una cuenta gratuita en https://github.com.
2. Presionar el botón **New** (nuevo repositorio).
3. Escribir un nombre, elegir si será **público** (cualquiera puede verlo) o **privado** (solo tú y quienes invites).
4. Presionar **Create repository**.

GitHub mostrará instrucciones con la dirección (URL) del repositorio, algo como:

```
https://github.com/tu-usuario/mi-proyecto.git
```

### 7.2 Autenticación

GitHub ya no acepta tu contraseña normal desde la terminal. Las opciones más simples son:

- Instalar **GitHub CLI** (`gh`) y ejecutar `gh auth login`, que guía todo el proceso.
- Usar **GitHub Desktop**, una aplicación gráfica que se encarga de la autenticación.
- Crear un **Personal Access Token** en la configuración de GitHub y usarlo en lugar de la contraseña.

### 7.3 Conectar tu repositorio local con GitHub

```bash
git remote add origin https://github.com/tu-usuario/mi-proyecto.git
git push -u origin main
```

- `remote add origin` le dice a Git: "existe una copia remota de este proyecto y la vamos a llamar `origin`". El nombre `origin` es una convención.
- `push` envía tus commits a GitHub.
- `-u` deja configurada la conexión, de modo que en el futuro basta con escribir `git push`.

### 7.4 Los cuatro comandos para trabajar con el remoto

| Comando | Qué hace | Analogía |
|---|---|---|
| `git clone URL` | Descarga un repositorio completo desde GitHub por primera vez | Descargar el álbum completo |
| `git push` | Envía tus commits nuevos a GitHub | Subir tus fotos nuevas |
| `git pull` | Trae a tu computador los commits nuevos que otras personas subieron | Descargar las fotos nuevas de los demás |
| `git fetch` | Consulta qué hay de nuevo en GitHub, pero sin mezclarlo todavía con tu trabajo | Mirar qué fotos nuevas hay sin descargarlas aún |

Regla práctica: **haz `git pull` antes de empezar a trabajar y antes de hacer `git push`.** Así reduces mucho la probabilidad de conflictos.

## 8. Colaborar en GitHub

### 8.1 Pull Request

Un **Pull Request** (PR) es una solicitud para que los cambios de una rama se incorporen a otra, normalmente a `main`. Es la forma estándar de colaborar.

El flujo es:

1. Creas una rama y haces tus cambios.
2. Subes la rama a GitHub con `git push -u origin nombre-rama`.
3. En GitHub, abres un Pull Request.
4. Otras personas revisan los cambios, comentan línea por línea y sugieren mejoras.
5. Cuando todos están de acuerdo, se presiona **Merge** y los cambios entran a `main`.

La ventaja es que nadie modifica la versión principal sin que otra persona haya revisado el trabajo.

### 8.2 Fork

Un **fork** es una copia de un repositorio ajeno en tu propia cuenta de GitHub. Se usa cuando quieres contribuir a un proyecto en el que no tienes permiso para escribir directamente (por ejemplo, un proyecto de código abierto).

El proceso habitual es: haces fork, clonas tu copia, haces cambios, los subes a tu copia y abres un Pull Request hacia el proyecto original.

Diferencia clave:

- **Clone**: copia un repositorio a tu computador.
- **Fork**: copia un repositorio a tu cuenta de GitHub.

### 8.3 Otras herramientas de GitHub

- **Issues**: una lista de tareas, errores reportados o ideas. Funcionan como tickets de seguimiento.
- **README.md**: archivo que GitHub muestra en la portada del repositorio. Explica de qué trata el proyecto y cómo usarlo.
- **GitHub Pages**: permite publicar un sitio web estático gratis directamente desde un repositorio.
- **GitHub Actions**: automatiza tareas, por ejemplo ejecutar pruebas cada vez que alguien sube cambios.

## 9. Flujo de trabajo típico de un día

```bash
git switch main
git pull                                # 1. traer lo último

git switch -c arreglo-formulario        # 2. crear una rama para tu tarea

# ... editar archivos ...

git status                              # 3. revisar qué cambió
git add .                               # 4. preparar cambios
git commit -m "Valida correo en formulario"   # 5. guardar

git push -u origin arreglo-formulario   # 6. subir la rama

# 7. abrir un Pull Request en GitHub, esperar revisión y hacer merge
```

## 10. El archivo .gitignore

Hay archivos que **no** deben guardarse en Git: contraseñas, archivos temporales, carpetas de dependencias que se pueden volver a descargar, etc.

Para eso se crea un archivo llamado `.gitignore` en la raíz del proyecto, con una regla por línea:

```
# Archivos de entorno con claves secretas
.env

# Dependencias que se reinstalan
node_modules/
venv/

# Archivos generados automáticamente
__pycache__/
*.log

# Archivos del sistema operativo
.DS_Store
Thumbs.db
```

Git ignorará todo lo que coincida con esas reglas. GitHub ofrece plantillas de `.gitignore` para cada lenguaje al crear un repositorio.

**Importante:** nunca subas contraseñas ni claves a GitHub, ni siquiera en repositorios privados. Si ocurre por error, considera esa clave comprometida y cámbiala de inmediato, porque queda en el historial.

## 11. Cómo deshacer cosas

| Situación | Comando |
|---|---|
| Quiero descartar cambios de un archivo que aún no preparé | `git restore archivo.py` |
| Agregué un archivo con `git add` por error | `git restore --staged archivo.py` |
| Quiero corregir el mensaje del último commit (aún no hecho push) | `git commit --amend -m "Nuevo mensaje"` |
| Quiero revertir un commit que ya subí a GitHub | `git revert ID_DEL_COMMIT` |

`git revert` es la forma segura de deshacer algo ya compartido: no borra el historial, sino que crea un commit nuevo que anula el anterior.

Ten cuidado con `git reset --hard`: borra cambios de forma definitiva. Úsalo solo cuando entiendas bien lo que hace.

## 12. Buenas prácticas

1. **Commits pequeños y frecuentes.** Cada commit debería representar un solo cambio lógico.
2. **Mensajes claros.** Describe qué hace el cambio, en tiempo presente: "Agrega", "Corrige", "Elimina".
3. **Una rama por tarea.** No trabajes directamente sobre `main`.
4. **Pull antes de push.** Mantén tu copia actualizada.
5. **Revisa antes de confirmar.** Usa `git status` y `git diff` antes de cada commit.
6. **Usa .gitignore desde el principio.**
7. **Escribe un README.** Tu yo del futuro lo agradecerá.

## 14. Resumen de comandos

```bash
# Configuración (una vez)
git config --global user.name "Tu Nombre"
git config --global user.email "tu@correo.com"

# Empezar
git init                        # crear repositorio nuevo
git clone URL                   # descargar repositorio existente

# Día a día
git status                      # ver estado
git add archivo                 # preparar un archivo
git add .                       # preparar todo
git commit -m "mensaje"         # guardar fotografía
git log --oneline               # ver historial
git diff                        # ver diferencias

# Ramas
git branch                      # listar ramas
git switch -c nombre            # crear y cambiar de rama
git switch nombre               # cambiar de rama
git merge nombre                # unir rama a la actual
git branch -d nombre            # borrar rama

# Remoto
git remote add origin URL       # conectar con GitHub
git push -u origin main         # primer envío
git push                        # enviar cambios
git pull                        # traer cambios

# Deshacer
git restore archivo             # descartar cambios no preparados
git restore --staged archivo    # sacar del área de preparación
git revert ID                   # revertir un commit de forma segura
```
